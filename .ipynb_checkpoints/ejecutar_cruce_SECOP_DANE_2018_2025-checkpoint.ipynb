## 🟤 CAPA BRONCE — Extracción y lectura de fuentes

### B1 — Leer SECOP II raw desde parquet local

```python
# ── B1: Leer SECOP II con DuckDB (bajo consumo de RAM) ───────────────────────
con = duckdb.connect(database=':memory:', config={'threads': 4, 'memory_limit': '6GB'})

print("⏳ Leyendo SECOP II raw...")
con.execute(f"""
    CREATE OR REPLACE TABLE bronze_secop AS
    SELECT *
    FROM read_parquet('{SECOP_RAW}')
""")

n_secop = con.execute("SELECT COUNT(*) FROM bronze_secop").fetchone()[0]
cols_secop = [r[0] for r in con.execute("DESCRIBE bronze_secop").fetchall()]

print(f"✅ SECOP cargado: {n_secop:,} registros | {len(cols_secop)} columnas")

# Verificar columnas clave
COLS_CLAVE = ['precio_base', 'valor_total_adjudicacion', 'divipola_municipio', 
              'anio', 'ordenentidad', 'fecha_de_publicacion_del']
print("\n🔍 Columnas clave:")
for c in COLS_CLAVE:
    estado = "✅" if c in cols_secop else "⚠️ FALTANTE"
    print(f"   {estado} {c}")
```

### B2 — Leer datos DANE locales (IPM, NBI, Etnia)

```python
# ── B2a: IPM (CNPV 2018) ─────────────────────────────────────────────────────
IPM_PATH   = "datos/dane_2018/ipm.xlsx"
NBI_PATH   = "datos/dane_2018/nbi_municipios_2018.parquet"
ETNIA_PATH = "datos/etnia_checkpoint.parquet"

print("⏳ Cargando fuentes DANE locales...")

# IPM
if os.path.exists(IPM_PATH):
    pdf_ipm = pd.read_excel(IPM_PATH)
    if 'anio' in pdf_ipm.columns:
        pdf_ipm = pdf_ipm.drop(columns=['anio'])
    pdf_ipm['divipola_municipio'] = pdf_ipm['divipola_municipio'].astype(str).str.zfill(5)
    con.register('bronze_ipm_df', pdf_ipm)
    con.execute("CREATE OR REPLACE TABLE bronze_ipm AS SELECT * FROM bronze_ipm_df")
    print(f"   ✅ IPM      : {len(pdf_ipm):,} filas | {pdf_ipm.shape[1]} columnas")
else:
    print(f"   ⚠️ IPM no encontrado: {IPM_PATH}")

# NBI
if os.path.exists(NBI_PATH):
    pdf_nbi = pd.read_parquet(NBI_PATH)
    if 'anio' in pdf_nbi.columns:
        pdf_nbi = pdf_nbi.drop(columns=['anio'])
    pdf_nbi['divipola_municipio'] = pdf_nbi['divipola_municipio'].astype(str).str.zfill(5)
    con.register('bronze_nbi_df', pdf_nbi)
    con.execute("CREATE OR REPLACE TABLE bronze_nbi AS SELECT * FROM bronze_nbi_df")
    print(f"   ✅ NBI      : {len(pdf_nbi):,} filas | {pdf_nbi.shape[1]} columnas")
else:
    print(f"   ⚠️ NBI no encontrado: {NBI_PATH}")

# Etnia
if os.path.exists(ETNIA_PATH):
    pdf_etnia = pd.read_parquet(ETNIA_PATH)
    if 'anio' in pdf_etnia.columns:
        pdf_etnia = pdf_etnia.drop(columns=['anio'])
    pdf_etnia['divipola_municipio'] = pdf_etnia['divipola_municipio'].astype(str).str.zfill(5)
    con.register('bronze_etnia_df', pdf_etnia)
    con.execute("CREATE OR REPLACE TABLE bronze_etnia AS SELECT * FROM bronze_etnia_df")
    print(f"   ✅ Etnia    : {len(pdf_etnia):,} filas | {pdf_etnia.shape[1]} columnas")
else:
    print(f"   ⚠️ Etnia no encontrada: {ETNIA_PATH}")
```

### B3 — Descargar DANE dinámicas vía API (EMICRON/CENU + TerriData)

```python
# ── B3: Descarga paginada de APIs datos.gov.co ────────────────────────────────

def fetch_api_all(url_base, nombre=""):
    """Descarga todos los registros de una API Socrata con paginación."""
    dfs, limit, offset = [], 50_000, 0
    print(f"   ⏳ Descargando {nombre}...", end="", flush=True)
    while True:
        url = f"{url_base}?$limit={limit}&$offset={offset}"
        try:
            resp = requests.get(url, timeout=60)
            resp.raise_for_status()
            data = resp.json()
            if not data:
                break
            dfs.append(pd.DataFrame(data))
            offset += limit
            print(".", end="", flush=True)
        except Exception as e:
            print(f"\n   ❌ Error en {nombre}: {e}")
            break
    print()
    return pd.concat(dfs, ignore_index=True) if dfs else pd.DataFrame()

# EMICRON / CENU
pdf_emicron = fetch_api_all("https://www.datos.gov.co/resource/r2bh-bfag.json", "EMICRON/CENU")
if not pdf_emicron.empty:
    pdf_emicron['divipola_municipio'] = pdf_emicron['divipola_municipio'].astype(str).str.zfill(5) \
        if 'divipola_municipio' in pdf_emicron.columns else '00000'
    if 'anio' not in pdf_emicron.columns:
        date_col = next((c for c in pdf_emicron.columns if 'fecha' in c.lower()), None)
        pdf_emicron['anio'] = (
            pd.to_datetime(pdf_emicron[date_col], errors='coerce').dt.year.fillna(2024).astype(int)
            if date_col else 2024
        )
    else:
        pdf_emicron['anio'] = pd.to_numeric(pdf_emicron['anio'], errors='coerce').fillna(0).astype(int)
    con.register('bronze_emicron_df', pdf_emicron)
    con.execute("CREATE OR REPLACE TABLE bronze_emicron AS SELECT * FROM bronze_emicron_df")
    print(f"   ✅ EMICRON  : {len(pdf_emicron):,} filas | {pdf_emicron.shape[1]} columnas")
else:
    print("   ⚠️ EMICRON vacío — se continuará sin esta fuente")
    con.execute("CREATE OR REPLACE TABLE bronze_emicron AS SELECT '' AS divipola_municipio, 0 AS anio LIMIT 0")

# TerriData
pdf_terridata = fetch_api_all("https://www.datos.gov.co/resource/64cq-xb2k.json", "TerriData")
if not pdf_terridata.empty:
    for col_name in ['código entidad territorial', 'codigo_entidad_territorial', 'Código Entidad Territorial']:
        if col_name in pdf_terridata.columns:
            pdf_terridata = pdf_terridata.rename(columns={col_name: 'divipola_municipio'})
            break
    if 'divipola_municipio' in pdf_terridata.columns:
        pdf_terridata['divipola_municipio'] = (
            pdf_terridata['divipola_municipio'].astype(str)
            .str.replace(r'\.0$', '', regex=True).str.zfill(5)
        )
    if 'anio' not in pdf_terridata.columns:
        year_col = next((c for c in pdf_terridata.columns 
                         if any(p in c.lower() for p in ['año', 'year', 'anio'])), None)
        date_col = next((c for c in pdf_terridata.columns if 'fecha' in c.lower()), None)
        if year_col:
            pdf_terridata['anio'] = pd.to_numeric(pdf_terridata[year_col], errors='coerce').fillna(0).astype(int)
        elif date_col:
            pdf_terridata['anio'] = pd.to_datetime(pdf_terridata[date_col], errors='coerce').dt.year.fillna(0).astype(int)
        else:
            pdf_terridata['anio'] = 2024
    else:
        pdf_terridata['anio'] = pd.to_numeric(pdf_terridata['anio'], errors='coerce').fillna(0).astype(int)
    con.register('bronze_terridata_df', pdf_terridata)
    con.execute("CREATE OR REPLACE TABLE bronze_terridata AS SELECT * FROM bronze_terridata_df")
    print(f"   ✅ TerriData: {len(pdf_terridata):,} filas | {pdf_terridata.shape[1]} columnas")
else:
    print("   ⚠️ TerriData vacío — se continuará sin esta fuente")
    con.execute("CREATE OR REPLACE TABLE bronze_terridata AS SELECT '' AS divipola_municipio, 0 AS anio LIMIT 0")

print("\n✅ Capa Bronce completa")
```

## ⚪ CAPA PLATA — Estandarización y cruces

### P1 — Limpiar y estandarizar SECOP

```python
# ── P1: Limpieza SECOP — trim, tipos numéricos, DIVIPOLA a 5 dígitos ─────────
print("⏳ Limpiando SECOP...")

con.execute("""
    CREATE OR REPLACE TABLE silver_secop AS
    SELECT DISTINCT
        -- DIVIPOLA normalizado a 5 dígitos
        LPAD(TRIM(CAST(divipola_municipio AS VARCHAR)), 5, '0')  AS divipola_municipio,

        -- Año
        TRY_CAST(anio AS INTEGER)                                AS anio,

        -- Flag nacional
        CASE WHEN TRIM(LOWER(COALESCE(ordenentidad,''))) = 'nacional'
             THEN TRUE ELSE FALSE END                            AS es_nacional,

        -- Valores monetarios
        TRY_CAST(precio_base              AS DOUBLE)             AS precio_base,
        TRY_CAST(valor_total_adjudicacion AS DOUBLE)             AS valor_total_adjudicacion,

        -- Resto de columnas: conservar todas (se incluyen abajo)
        * EXCLUDE (divipola_municipio, anio, es_nacional,
                   precio_base, valor_total_adjudicacion)
    FROM bronze_secop
    WHERE divipola_municipio IS NOT NULL
""")

n_silver = con.execute("SELECT COUNT(*) FROM silver_secop").fetchone()[0]
n_nac    = con.execute("SELECT COUNT(*) FROM silver_secop WHERE es_nacional = TRUE").fetchone()[0]
n_ter    = con.execute("SELECT COUNT(*) FROM silver_secop WHERE es_nacional = FALSE").fetchone()[0]

print(f"✅ SECOP limpio : {n_silver:,} registros")
print(f"   🏛️  Nacional    : {n_nac:,}")
print(f"   🏘️  Territorial : {n_ter:,}")
```

### P2 — Estandarizar fuentes DANE

```python
# ── P2: Estandarizar DIVIPOLA en todas las tablas DANE ───────────────────────
print("⏳ Estandarizando fuentes DANE...")

# Función auxiliar: renombrar columnas (excepto claves) con sufijo
def estandarizar_dane(con, tabla_src, tabla_dst, sufijo, keys=('divipola_municipio',)):
    cols_raw = [r[0] for r in con.execute(f"DESCRIBE {tabla_src}").fetchall()]
    selects = []
    for c in cols_raw:
        if c == 'divipola_municipio':
            selects.append(f"LPAD(TRIM(CAST({c} AS VARCHAR)), 5, '0') AS divipola_municipio")
        elif c == 'anio':
            selects.append(f"TRY_CAST({c} AS INTEGER) AS anio")
        elif c in keys:
            selects.append(c)
        else:
            selects.append(f"{c} AS {c}_{sufijo}")
    sql = f"CREATE OR REPLACE TABLE {tabla_dst} AS SELECT {', '.join(selects)} FROM {tabla_src}"
    con.execute(sql)
    n = con.execute(f"SELECT COUNT(*) FROM {tabla_dst}").fetchone()[0]
    print(f"   ✅ {tabla_dst} : {n:,} filas")

# Solo tablas que existen con datos
for src, dst, sufijo, keys in [
    ('bronze_ipm',       'silver_ipm',       'ipm',       ('divipola_municipio',)),
    ('bronze_nbi',       'silver_nbi',       'nbi',       ('divipola_municipio',)),
    ('bronze_etnia',     'silver_etnia',     'etnia',     ('divipola_municipio',)),
    ('bronze_emicron',   'silver_emicron',   'emicron',   ('divipola_municipio', 'anio')),
    ('bronze_terridata', 'silver_terridata', 'terridata', ('divipola_municipio', 'anio')),
]:
    try:
        estandarizar_dane(con, src, dst, sufijo, keys)
    except Exception as e:
        print(f"   ⚠️ {src} no disponible: {e}")

print("\n✅ Fuentes DANE estandarizadas")
```

### P3 — Cruzar SECOP con DANE y separar Nacional / Territorial

```python
# ── P3: LEFT JOINs y partición Nacional / Territorial ────────────────────────
print("⏳ Ejecutando cruces SECOP × DANE...")

# Verificar qué tablas de plata están disponibles
tablas_silver = {r[0] for r in con.execute("SHOW TABLES").fetchall()}

join_ipm       = "LEFT JOIN silver_ipm      si  ON s.divipola_municipio = si.divipola_municipio" \
                 if 'silver_ipm'       in tablas_silver else ""
join_nbi       = "LEFT JOIN silver_nbi      sn  ON s.divipola_municipio = sn.divipola_municipio" \
                 if 'silver_nbi'       in tablas_silver else ""
join_etnia     = "LEFT JOIN silver_etnia    se  ON s.divipola_municipio = se.divipola_municipio" \
                 if 'silver_etnia'     in tablas_silver else ""
join_emicron   = "LEFT JOIN silver_emicron  sem ON s.divipola_municipio = sem.divipola_municipio AND s.anio = sem.anio" \
                 if 'silver_emicron'   in tablas_silver else ""
join_terridata = "LEFT JOIN silver_terridata st  ON s.divipola_municipio = st.divipola_municipio  AND s.anio = st.anio" \
                 if 'silver_terridata' in tablas_silver else ""

sql_cruce = f"""
    CREATE OR REPLACE TABLE silver_cruce AS
    SELECT s.*
    FROM silver_secop s
    {join_ipm}
    {join_nbi}
    {join_etnia}
    {join_emicron}
    {join_terridata}
"""
con.execute(sql_cruce)

# Partición
con.execute("""
    CREATE OR REPLACE TABLE silver_nacional AS
    SELECT * FROM silver_cruce WHERE es_nacional = TRUE
""")
con.execute("""
    CREATE OR REPLACE TABLE silver_territorial AS
    SELECT * FROM silver_cruce WHERE es_nacional = FALSE
""")

n_cruce = con.execute("SELECT COUNT(*) FROM silver_cruce").fetchone()[0]
n_nac   = con.execute("SELECT COUNT(*) FROM silver_nacional").fetchone()[0]
n_ter   = con.execute("SELECT COUNT(*) FROM silver_territorial").fetchone()[0]

print(f"✅ Cruce total      : {n_cruce:,}")
print(f"   🏛️  Nacional     : {n_nac:,}")
print(f"   🏘️  Territorial  : {n_ter:,}")
print("\n✅ Capa Plata completa")
```

## 🟡 CAPA ORO — Detección y tratamiento de atípicos

### O1 — Función de detección (IF + LOF + SVM, votación ≥ 2/3)

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM
from sklearn.impute import KNNImputer

def detect_and_treat_outliers(pdf: pd.DataFrame) -> pd.DataFrame:
    """
    Detección por votación (IF + LOF + SVM) y tratamiento:
      1. Winsorización P5–P95
      2. KNN Imputer si persiste la anomalía
    """
    # Valores por defecto cuando no hay suficientes datos
    defaults = {
        'atipico_isolation_forest': False, 'score_isolation_forest': 1.0,
        'atipico_lof':              False, 'score_lof':              1.0,
        'atipico_svm':              False, 'score_svm':              1.0,
        'es_atipico': False, 'tipo_atipico': 'normal', 'metodo_tratamiento': 'sin_tratamiento',
        'valor_tratado_precio_base':   pdf['precio_base'],
        'valor_tratado_adjudicacion':  pdf['valor_total_adjudicacion'],
    }
    if len(pdf) < 5:
        return pdf.assign(**defaults)

    feats = ['precio_base', 'valor_total_adjudicacion']
    X = pdf[feats].fillna(0).values

    # ── Modelos ──────────────────────────────────────────────────────────────
    if_model  = IsolationForest(contamination=0.05, random_state=42)
    lof_model = LocalOutlierFactor(n_neighbors=min(20, len(pdf) - 1), contamination=0.05)
    svm_model = OneClassSVM(kernel='rbf', nu=0.05)

    preds_if  = if_model.fit_predict(X);      scores_if  = if_model.decision_function(X)
    preds_lof = lof_model.fit_predict(X);     scores_lof = lof_model.negative_outlier_factor_
    preds_svm = svm_model.fit_predict(X);     scores_svm = svm_model.decision_function(X)

    pdf = pdf.copy()
    pdf['atipico_isolation_forest'] = (preds_if  == -1)
    pdf['score_isolation_forest']   = scores_if
    pdf['atipico_lof']              = (preds_lof == -1)
    pdf['score_lof']                = scores_lof
    pdf['atipico_svm']              = (preds_svm == -1)
    pdf['score_svm']                = scores_svm

    # ── Votación ─────────────────────────────────────────────────────────────
    votos = (pdf['atipico_isolation_forest'].astype(int) +
             pdf['atipico_lof'].astype(int) +
             pdf['atipico_svm'].astype(int))
    pdf['es_atipico'] = (votos >= 2)

    med_pb = pdf['precio_base'].median()
    med_va = pdf['valor_total_adjudicacion'].median()
    pdf['tipo_atipico'] = pdf.apply(
        lambda r: 'normal' if not r['es_atipico']
                  else ('alto' if r['precio_base'] > med_pb or r['valor_total_adjudicacion'] > med_va
                        else 'bajo'), axis=1
    )

    # ── Tratamiento ───────────────────────────────────────────────────────────
    pdf['valor_tratado_precio_base']  = pdf['precio_base']
    pdf['valor_tratado_adjudicacion'] = pdf['valor_total_adjudicacion']
    pdf['metodo_tratamiento']         = 'sin_tratamiento'

    is_out = pdf['es_atipico']
    p05_pb, p95_pb = pdf['precio_base'].quantile([0.05, 0.95])
    p05_va, p95_va = pdf['valor_total_adjudicacion'].quantile([0.05, 0.95])

    pdf.loc[is_out, 'valor_tratado_precio_base']  = np.clip(pdf.loc[is_out, 'precio_base'],              p05_pb, p95_pb)
    pdf.loc[is_out, 'valor_tratado_adjudicacion'] = np.clip(pdf.loc[is_out, 'valor_total_adjudicacion'], p05_va, p95_va)
    pdf.loc[is_out, 'metodo_tratamiento']         = 'winsorizado'

    # KNN para los que persisten atípicos después de winsorizar
    X2 = pdf[['valor_tratado_precio_base', 'valor_tratado_adjudicacion']].values
    still_out = is_out & (if_model.fit_predict(X2) == -1)
    if still_out.sum() > 0:
        pdf.loc[still_out, 'valor_tratado_precio_base']  = np.nan
        pdf.loc[still_out, 'valor_tratado_adjudicacion'] = np.nan
        pdf.loc[still_out, 'metodo_tratamiento']         = 'imputado_knn'
        imputed = KNNImputer(n_neighbors=5).fit_transform(
            pdf[['valor_tratado_precio_base', 'valor_tratado_adjudicacion']]
        )
        pdf['valor_tratado_precio_base']  = imputed[:, 0]
        pdf['valor_tratado_adjudicacion'] = imputed[:, 1]

    return pdf

print("✅ Función detect_and_treat_outliers definida")
```

### O2 — Aplicar detección por grupo (ordenentidad × anio)

```python
# ── O2: Aplicar detección de atípicos por grupo ───────────────────────────────
# Se trabaja en pandas (DuckDB exporta a pandas, procesamos, reimportamos)

def aplicar_capa_oro(tabla_silver, nombre):
    print(f"\n⏳ [{nombre}] Exportando a pandas...")
    pdf = con.execute(f"SELECT * FROM {tabla_silver}").df()

    # Rellenar nulos en columnas clave de agrupación
    pdf['ordenentidad'] = pdf['ordenentidad'].fillna('Desconocido') \
        if 'ordenentidad' in pdf.columns else 'Desconocido'
    pdf['anio'] = pdf['anio'].fillna(2018).astype(int)
    pdf['precio_base']              = pd.to_numeric(pdf['precio_base'],              errors='coerce').fillna(0)
    pdf['valor_total_adjudicacion'] = pd.to_numeric(pdf['valor_total_adjudicacion'], errors='coerce').fillna(0)

    print(f"   📦 {len(pdf):,} registros — procesando por grupo (ordenentidad × anio)...")
    grupos = pdf.groupby(['ordenentidad', 'anio'], group_keys=False)
    total_grupos = grupos.ngroups

    resultados = []
    for i, (key, grupo) in enumerate(grupos, 1):
        resultados.append(detect_and_treat_outliers(grupo.copy()))
        if i % max(1, total_grupos // 10) == 0 or i == total_grupos:
            print(f"   ✅ {i}/{total_grupos} grupos procesados", end="\r")

    df_gold = pd.concat(resultados, ignore_index=True)
    n_atip  = df_gold['es_atipico'].sum()
    print(f"\n   📊 Atípicos detectados: {n_atip:,} ({n_atip/len(df_gold)*100:.2f}%)")
    return df_gold

df_gold_nac = aplicar_capa_oro('silver_nacional',    'Nacional')
df_gold_ter = aplicar_capa_oro('silver_territorial', 'Territorial')

print("\n✅ Capa Oro completa")
```

## 💾 Paso 7 — Guardar resultados en `datos/processed/`

```python
# ── Paso 7: Guardar los tres parquets de salida ───────────────────────────────
import pyarrow as pa
import pyarrow.parquet as pq

OUT_DIR = "datos/processed"
os.makedirs(OUT_DIR, exist_ok=True)

def guardar_parquet(df, path, nombre):
    table = pa.Table.from_pandas(df, preserve_index=False)
    pq.write_table(table, path, compression='snappy')
    size_mb = os.path.getsize(path) / 1_048_576
    print(f"   💾 {nombre}")
    print(f"      Ruta   : {path}")
    print(f"      Filas  : {len(df):,}")
    print(f"      Tamaño : {size_mb:.1f} MB")

print("⏳ Guardando archivos...\n")

# Nacional
guardar_parquet(
    df_gold_nac,
    f"{OUT_DIR}/cruce_secop_dane_nacional_2018_2025.parquet",
    "cruce_secop_dane_nacional_2018_2025.parquet"
)

# Territorial
guardar_parquet(
    df_gold_ter,
    f"{OUT_DIR}/cruce_secop_dane_territorial_2018_2025.parquet",
    "cruce_secop_dane_territorial_2018_2025.parquet"
)

# Reporte de atípicos (unión de ambos)
COLS_REPORTE = [
    'divipola_municipio', 'anio', 'precio_base', 'valor_total_adjudicacion',
    'valor_tratado_precio_base', 'valor_tratado_adjudicacion',
    'atipico_isolation_forest', 'atipico_lof', 'atipico_svm', 'metodo_tratamiento'
]

df_atip_nac = df_gold_nac[df_gold_nac['es_atipico'] == True]
df_atip_ter = df_gold_ter[df_gold_ter['es_atipico'] == True]
df_reporte  = pd.concat([df_atip_nac, df_atip_ter], ignore_index=True)
cols_rep    = [c for c in COLS_REPORTE if c in df_reporte.columns]
df_reporte  = df_reporte[cols_rep]

guardar_parquet(
    df_reporte,
    f"{OUT_DIR}/reporte_atipicos_2018_2025.parquet",
    "reporte_atipicos_2018_2025.parquet"
)

print("\n✅ Pipeline ejecutado exitosamente")
print(f"\n📁 Archivos en: {os.path.abspath(OUT_DIR)}/")
```

## ✅ Verificación rápida de los archivos generados

```python
# ── Verificación de salida ────────────────────────────────────────────────────
archivos = {
    'Nacional'    : f"{OUT_DIR}/cruce_secop_dane_nacional_2018_2025.parquet",
    'Territorial' : f"{OUT_DIR}/cruce_secop_dane_territorial_2018_2025.parquet",
    'Atípicos'    : f"{OUT_DIR}/reporte_atipicos_2018_2025.parquet",
}

print("=" * 65)
print("  VERIFICACIÓN DE ARCHIVOS GENERADOS")
print("=" * 65)

for nombre, path in archivos.items():
    if os.path.exists(path):
        df_v = pd.read_parquet(path)
        size_mb = os.path.getsize(path) / 1_048_576
        n_atip  = df_v['es_atipico'].sum() if 'es_atipico' in df_v.columns else '-'
        print(f"\n  ✅ {nombre}")
        print(f"     Filas      : {len(df_v):,}")
        print(f"     Columnas   : {df_v.shape[1]}")
        print(f"     Atípicos   : {n_atip}")
        print(f"     Tamaño     : {size_mb:.1f} MB")
        display(df_v.head(3))
    else:
        print(f"\n  ❌ {nombre} — archivo no encontrado: {path}")
```
