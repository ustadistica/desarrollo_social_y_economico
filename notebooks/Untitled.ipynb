```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.ticker import FuncFormatter
import warnings
warnings.filterwarnings('ignore')

# ── Rutas de los parquets generados por el pipeline ──────────────────────────
PATH_NAC  = "datos/processed/cruce_secop_dane_nacional_2018_2025.parquet"
PATH_TER  = "datos/processed/cruce_secop_dane_territorial_2018_2025.parquet"
PATH_ATIP = "datos/processed/reporte_atipicos_2018_2025.parquet"

df_nac  = pd.read_parquet(PATH_NAC)
df_ter  = pd.read_parquet(PATH_TER)
df_atip = pd.read_parquet(PATH_ATIP)

# Vista combinada (útil para análisis globales)
df_nac["origen"]  = "Nacional"
df_ter["origen"]  = "Territorial"
df_all = pd.concat([df_nac, df_ter], ignore_index=True)

print(f"Nacional   : {df_nac.shape[0]:,} filas × {df_nac.shape[1]} columnas")
print(f"Territorial: {df_ter.shape[0]:,} filas × {df_ter.shape[1]} columnas")
print(f"Combinado  : {df_all.shape[0]:,} filas × {df_all.shape[1]} columnas")
print(f"Atípicos   : {df_atip.shape[0]:,} filas × {df_atip.shape[1]} columnas")
```
