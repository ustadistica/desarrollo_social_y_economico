4. Distribución de la inversión



```python
# ============================================================================
# 4. DISTRIBUCIÓN DE LA INVERSIÓN MUNICIPAL (PROFESIONAL SIN SEABORN)
# ============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# -------------------------------
# 1. CONFIGURACIÓN VISUAL
# -------------------------------
plt.style.use("ggplot")
plt.rcParams["figure.dpi"] = 120
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
plt.rcParams["font.size"] = 10

print("=" * 70)
print("DISTRIBUCIÓN DE LA INVERSIÓN MUNICIPAL")
print("=" * 70)

# -------------------------------
# 2. PREPARACIÓN DE DATOS
# -------------------------------
data_monto = df["monto_total_contratos"].fillna(0)
data_contratos = df["num_contratos"].fillna(0)

monto_no_cero = data_monto[data_monto > 0]
df_monto_pos = df[df["monto_total_contratos"] > 0].copy()
df_monto_pos["log_monto"] = np.log10(df_monto_pos["monto_total_contratos"])

# -------------------------------
# 3. FUNCIONES AUXILIARES
# -------------------------------
def formato_moneda(x, pos=None):
    if x >= 1e12:
        return f"${x/1e12:.1f}T"
    elif x >= 1e9:
        return f"${x/1e9:.1f}B"
    elif x >= 1e6:
        return f"${x/1e6:.1f}M"
    elif x >= 1e3:
        return f"${x/1e3:.0f}K"
    return f"${x:,.0f}"

def kde_manual(data, points=300):
    """
    Estimación de densidad simple sin seaborn ni scipy.
    """
    data = np.asarray(data)
    data = data[np.isfinite(data)]
    
    if len(data) < 2:
        return None, None
    
    x_grid = np.linspace(data.min(), data.max(), points)
    bandwidth = 1.06 * data.std() * (len(data) ** (-1/5)) if data.std() > 0 else 1
    
    density = np.zeros_like(x_grid)
    for x in data:
        density += np.exp(-0.5 * ((x_grid - x) / bandwidth) ** 2)
    
    density /= (len(data) * bandwidth * np.sqrt(2 * np.pi))
    return x_grid, density

# -------------------------------
# 4. ESTADÍSTICAS CLAVE
# -------------------------------
mean_monto = data_monto.mean()
median_monto = data_monto.median()
q1_monto = data_monto.quantile(0.25)
q3_monto = data_monto.quantile(0.75)
p90_monto = data_monto.quantile(0.90)
p95_monto = data_monto.quantile(0.95)

mean_contratos = data_contratos.mean()
median_contratos = data_contratos.median()

sin_contratos = (data_monto == 0).sum()
con_contratos = (data_monto > 0).sum()
porc_sin = (sin_contratos / len(df)) * 100

# -------------------------------
# 5. CREAR FIGURA
# -------------------------------
fig, axes = plt.subplots(2, 2, figsize=(16, 11))
fig.suptitle(
    "Análisis de la distribución de la contratación municipal",
    fontsize=16,
    fontweight="bold",
    y=1.02
)

# ============================================================================
# 5.1 HISTOGRAMA DEL MONTO TOTAL CONTRATADO
# ============================================================================
ax1 = axes[0, 0]

counts, bins, _ = ax1.hist(
    data_monto,
    bins=40,
    color="#4C78A8",
    edgecolor="black",
    alpha=0.75
)

ax1.axvline(mean_monto, color="red", linestyle="--", linewidth=2,
            label=f"Media: {formato_moneda(mean_monto)}")
ax1.axvline(median_monto, color="green", linestyle="--", linewidth=2,
            label=f"Mediana: {formato_moneda(median_monto)}")
ax1.axvline(p90_monto, color="purple", linestyle=":", linewidth=2,
            label=f"P90: {formato_moneda(p90_monto)}")

ax1.set_title("Distribución del monto total contratado", fontweight="bold")
ax1.set_xlabel("Monto total contratado")
ax1.set_ylabel("Número de municipios")
ax1.xaxis.set_major_formatter(FuncFormatter(formato_moneda))
ax1.legend()

ax1.text(
    0.98, 0.95,
    f"Municipios sin contratos: {sin_contratos:,}\n({porc_sin:.1f}% del total)",
    transform=ax1.transAxes,
    ha="right", va="top",
    fontsize=9,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#FFF3CD", edgecolor="#D6B656")
)

# ============================================================================
# 5.2 HISTOGRAMA DEL NÚMERO DE CONTRATOS
# ============================================================================
ax2 = axes[0, 1]

ax2.hist(
    data_contratos,
    bins=40,
    color="#F28E2B",
    edgecolor="black",
    alpha=0.75
)

ax2.axvline(mean_contratos, color="red", linestyle="--", linewidth=2,
            label=f"Media: {mean_contratos:.1f}")
ax2.axvline(median_contratos, color="green", linestyle="--", linewidth=2,
            label=f"Mediana: {median_contratos:.0f}")

ax2.set_title("Distribución del número de contratos", fontweight="bold")
ax2.set_xlabel("Número de contratos")
ax2.set_ylabel("Número de municipios")
ax2.legend()

# ============================================================================
# 5.3 DISTRIBUCIÓN EN ESCALA LOGARÍTMICA
# ============================================================================
ax3 = axes[1, 0]

ax3.hist(
    df_monto_pos["log_monto"],
    bins=35,
    color="#59A14F",
    edgecolor="black",
    alpha=0.75
)

ax3.axvline(df_monto_pos["log_monto"].mean(), color="red", linestyle="--", linewidth=2,
            label=f"Media log10: {df_monto_pos['log_monto'].mean():.2f}")
ax3.axvline(df_monto_pos["log_monto"].median(), color="green", linestyle="--", linewidth=2,
            label=f"Mediana log10: {df_monto_pos['log_monto'].median():.2f}")

ax3.set_title("Distribución del monto contratado (escala log10)", fontweight="bold")
ax3.set_xlabel("log10(Monto total contratado)")
ax3.set_ylabel("Número de municipios")
ax3.legend()

ax3.text(
    0.98, 0.95,
    "La escala logarítmica permite\nvisualizar mejor la dispersión\nentre municipios pequeños y grandes.",
    transform=ax3.transAxes,
    ha="right", va="top",
    fontsize=9,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#E8F5E9", edgecolor="#81C784")
)

# ============================================================================
# 5.4 BOXPLOT PROFESIONAL
# ============================================================================
ax4 = axes[1, 1]

bp = ax4.boxplot(
    monto_no_cero,
    vert=True,
    patch_artist=True,
    widths=0.5,
    showfliers=True,
    boxprops=dict(facecolor="#A0CBE8", color="#1F4E79", linewidth=1.5),
    medianprops=dict(color="darkred", linewidth=2),
    whiskerprops=dict(color="#1F4E79", linewidth=1.5),
    capprops=dict(color="#1F4E79", linewidth=1.5),
    flierprops=dict(marker='o', markersize=4, alpha=0.5, markerfacecolor='gray')
)

ax4.set_title("Boxplot del monto contratado\n(solo municipios con contratos)", fontweight="bold")
ax4.set_ylabel("Monto total contratado")
ax4.set_xticks([1])
ax4.set_xticklabels(["Monto > 0"])
ax4.set_yscale("log")
ax4.yaxis.set_major_formatter(FuncFormatter(formato_moneda))

ax4.text(
    0.98, 0.95,
    f"Q1: {formato_moneda(q1_monto)}\n"
    f"Q3: {formato_moneda(q3_monto)}\n"
    f"P95: {formato_moneda(p95_monto)}",
    transform=ax4.transAxes,
    ha="right", va="top",
    fontsize=9,
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#E3F2FD", edgecolor="#64B5F6")
)

# -------------------------------
# 6. AJUSTE FINAL
# -------------------------------
plt.tight_layout()
plt.show()

# ============================================================================
# 7. RESUMEN EJECUTIVO
# ============================================================================
print("\n📊 RESUMEN EJECUTIVO DE DISTRIBUCIÓN")
print("-" * 70)

print(f"• Total de municipios analizados: {len(df):,}")
print(f"• Municipios sin contratos: {sin_contratos:,} ({porc_sin:.1f}%)")
print(f"• Municipios con contratos: {con_contratos:,}")

print("\n💰 Monto contratado")
print(f"• Monto total contratado: {formato_moneda(data_monto.sum())}")
print(f"• Promedio por municipio (incluye ceros): {formato_moneda(mean_monto)}")
print(f"• Mediana por municipio: {formato_moneda(median_monto)}")
print(f"• Promedio por municipio (solo con contratos): {formato_moneda(monto_no_cero.mean()) if len(monto_no_cero) > 0 else 'N/A'}")
print(f"• Percentil 90: {formato_moneda(p90_monto)}")
print(f"• Percentil 95: {formato_moneda(p95_monto)}")

print("\n📑 Número de contratos")
print(f"• Total de contratos: {data_contratos.sum():,.0f}")
print(f"• Promedio de contratos por municipio: {mean_contratos:.1f}")
print(f"• Mediana de contratos por municipio: {median_contratos:.0f}")

if data_monto.sum() > 0:
    top_10_share = data_monto.sort_values(ascending=False).head(10).sum() / data_monto.sum() * 100
    print("\n📌 Concentración")
    print(f"• Los 10 municipios con mayor contratación concentran el {top_10_share:.1f}% del monto total contratado.")
```

```python
# ============================================================================
# 5. DESIGUALDAD DE LA INVERSIÓN: COEFICIENTE DE GINI
# ============================================================================

def gini(array):
    array = np.array(array, dtype=np.float64)
    array = array[array >= 0]
    
    if len(array) == 0 or array.sum() == 0:
        return np.nan
    
    array = np.sort(array)
    n = len(array)
    acumulado = np.cumsum(array)
    
    gini_coef = (n + 1 - 2 * np.sum(acumulado) / acumulado[-1]) / n
    return gini_coef

gini_monto = gini(data_monto)

print("=" * 70)
print("DESIGUALDAD EN LA DISTRIBUCIÓN DE LA CONTRATACIÓN")
print("=" * 70)
print(f"Coeficiente de Gini del monto contratado: {gini_monto:.3f}")
```
